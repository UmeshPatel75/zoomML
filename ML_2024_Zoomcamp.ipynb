```python
import requests

# URL of the file to download
url = "https://raw.githubusercontent.com/alexeygrigorev/datasets/master/laptops.csv"

# Send a GET request to the URL
response = requests.get(url)

# Open a file in write mode and save the content
with open("laptops.csv", "wb") as file:
    file.write(response.content)  # Write the content to the file

# Now the file 'laptops.csv' is downloaded and saved in the current directory
```
